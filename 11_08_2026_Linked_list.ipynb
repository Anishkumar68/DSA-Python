```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None


class Linked_list:
    def __init__(self):
        # values those we insert in head 
        self.head = None
        # len current ll == 0
        self.n = 0

    
    # get len of Linked_list
    def __len__(self):
        # ll += len(ll)
        return self.n 

    # insert from head 
    def insert_head(self, value):
        # new node
        new_node = Node(value)
        # pointer head values == next node value
        new_node.next = self.head
        # insert new_node value in head
        self.head = new_node

        # len increment
        self.n +=1

    # traverse ll
    def __str__(self):
        # current head = index => 0
        current = self.head

        # empty str
        result = ''
        # condition: if ll len !=0
        while current != None:
            # adding 4->
            result = result + str(current.data) + '->'
            # print(current.data)
            current = current.next

        # 4->3-> to stop last -> this we reverse
        return result[:-2]

    def append(self, value):
        # inheretance value from node class 
        new_node = Node(value)


        # if ll = 0 print(node current value)
        if self.head == None:
            self.head = new_node
            self.n += 1
            return

        # current ll index 0 value
        current = self.head
        # condition: if ll len !=0
        while current.next != None:
            current = current.next

        current.next = new_node
        self.n +=1

    def insert_after(self, value, after):
        # create new node 
        new_node = Node(value)

        current = self.head

        while current != None:
            if current.data == after:
                new_node.next = current.next
                current.next = new_node 

                self.n += 1

                return

            current = current.next

            

    def insert_at(self,index, value):
            #  if head == none then insert at head == value eg. 22 at index 0
            if index == 0:
                self.insert_head(value)
                return

            # create new node 
            new_node = Node(value)
            current = self.head

            for i in range(index - 1):
                current = current.next

            new_node.next = current.next
            current.next = new_node

            self.n +=1

    def delete_head(self):
        if self.head == None:
            return

        # move head to the second node
        self.head = self.head.next
        # decrese ll nodes count 
        self.n -=1

    def pop(self):
        if self.head == None:
            # empty ll
            return

        # when single item left in ll single node
        # if self.head.next != None:
        #     self.head = None 
        #     self.n -=1
    
        # method 2  
        if current == None:
            # when only one item its not the next its head so we delete head in that case so we call our already declared delete_head()
            self.delete_head()

        # second lastnode
        current = self.head

        # while the end is node value == none remove the second last means 5->None so remove the 5
        while current.next.next != None:
            current = current.next

        current.next = None
        self.n -=1
        
    def remove(self, value):
        # check head empty 
        if self.head == None:
            return 

        # if not check does value is in head
        if self.head.data == value:
            return self.delete_head()

        # current step 
        current = self.head

        # loop to check current value if match remove it 
        while current.next != None:
            if current.next.data == value:
                break

            current =current.next

        # got the values in ll 
        if current.next ==None:
            return "404: Not found"
        # values didn't found 
        else:
            current.next = current.next.next
            self.n -=1

    def delete_at(self,index):
        if self.head == None:
            return

        if index == 0:
            self.delete_head()

        current = self.head

        for _ in range( index -1 ):
            if current.next == None:
                current = current.next

        current.next = current.next.next
        self.n -=1

    def search_in(self, item):
        # check empty 
        if self.head == None:
            return 

        # assing current
        current = self.head
        #  position for check 
        pos = 0
        # traverse
        while current !=None:
            if current.data == item:
                # return the item position 
                return pos
            # Move to the requested index
            current = current.next
            #move to next node
            pos +=1

        # if not found value 
        return "Not Found"

    def __getitem__(self,index)
        current = self.head
        pos = 0 
        while current != None:
            if pos == index:
                return current.data

            current = current.next 
            pos +=1


        return "IndexError"
    
    def clear(self):
        self.head = None
        self.n = 0

#  call linked list obj
ll = Linked_list()

# by loop we adding node value
for i in range(0,100,7):
    ll.insert_head(i)

# assign nodes
a= Node(5)
b= Node(15) 
c = Node(30)

# connect with next node a->b->c-> null
a.next = b
b.next = c   

#  Len of LL - no of nodes
len(ll)
```
