```python
from calendar import c
from multiprocessing import current_process


class dictionary:
 
    def __init__(self, size : int) -> int:
        # Get the total capacity of the hash table 
        self.size = size
        # Create an array to store the VALUES, initialized with None
        self.data = [None] * self.size
        # Create a parallel array to store the KEYS, initialized with None
        self.slots = [None] * self.size

    # Insert or Update data in the hashtable 
    def put(self, key, value):
        # Step 1: Calculate the initial index using the hash function
        hash_value = self._hashfunction(key)

        # Step 2: Check if the calculated slot is empty
        if self.slots[hash_value] is None:
            # It's empty! Insert the key and value here.
            self.slots[hash_value] = key
            self.data[hash_value] = value
        else: 
            # Step 3: The slot is occupied. Is it the SAME key?
            if self.slots[hash_value] == key:
                # Yes, it's an update. Just overwrite the old value.
                self.data[hash_value] = value
            else:
                # Step 4: COLLISION! It's a different key. 
                # We need to find the next empty slot using Linear Probing.
                new_hash_value = self.rehash(hash_value)
                
                # Keep probing until we find an empty slot OR the existing key
                while self.slots[new_hash_value] != None and self.slots[new_hash_value] != key:
                    new_hash_value = self.rehash(new_hash_value)

                # Now that we found the right spot, insert or update
                if self.slots[new_hash_value] == None:
                    # Found an empty slot, insert new key/value
                    self.slots[new_hash_value] = key
                    self.data[new_hash_value] = value
                else:
                    # Found the existing key, update the value
                    self.data[new_hash_value] = value
    def __setitem__(self, key, value):
        self.put(key,value)

    def get(self,key):
        # start position take hashfunction value 
        start_position = self._hashfunction(key)
        # current position the index 
        current_position = start_position
        # check the hash table if current position value is not none so find the value for that we are seeking
        while self.slots[current_position] != None:
            if self.slots[current_position] == key:
                return self.data[current_position]

            current_position = self.rehash(current_position)

            if current_position == start_position:
                return "we are start poition so we Not found"

        return "None Not found"

    def __getitem__(self, key):
        return self.get(key)
        
    def rehash(self, old_hash):
        # Linear Probing: Move to the next index.
        # We use modulo (%) by self.size to wrap around to the start 
        # if we reach the end of the array.
        return (old_hash + 1) % self.size

    def _hashfunction(self, key):
        # Convert the key to a hash integer, make it positive, 
        # and use modulo to ensure it fits within the array size.
        return abs(hash(key)) % self.size


d = dictionary(5)


```
