```python
class Node:
    def __init__(self, value):
        self.data = value 
        self.next = None

class Stack:

    def __init__(self):
        self.top = None
        self.height = 0

    # check the stack full or empty [bool: true, False]
    def is_empty(self):
        return self.top == None

    # Add value in stack from top
    def push(self, value):
        new_node = Node(value)
        new_node.next = self.top
        self.top = new_node
        self.height +=1

    # peek 
    def peek(self):
        if self.is_empty():
            return "Stack empty"
        else:
            self.top = self.top.data
            return self.top

    # top item remove 
    def pop(self):
        if self.is_empty():
            return "stack empty"
        else:
            self.top = self.top.next
            self.height -=1
         

    # traverse
    def traverse(self):
        temp =self.top
        while temp!= None:
            print(temp.data)
            temp = temp.next 



st = Stack()
for i in range(0,10):
    st.push(i)
```

```python
# traverse()
st.traverse()
# pop
st.pop() 
# st.traverse()

# peek
st.peek()
```
