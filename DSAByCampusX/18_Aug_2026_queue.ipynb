# Queue


```python
class Node:

    def __init__(self, value):
        self.data = value
        self.next = None
    
class Queue:
    def __init__(self):
        self.front = None   # Points to the first node
        self.rear = None    # Points to the last node
        self.height = 0

    def isempty(self):
        return self.front == None

    def enqueue(self, value):
        new_node = Node(value)

        if self.isempty():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.height += 1

    # remove value from queue
    def dequeue(self):
        if self.isempty():
            return "Queue is empty"

        self.front = self.front.next

        if self.rear is None:
            return self.rear is None
      

        self.height -=1

    # PEEK

    def peek(self):
        if self.isempty():
            return "queue is empty"

        return self.front.data


    # traverse
    def traverse(self):
        temp = self.front 
        while temp != None:
            print(temp.data)
            temp = temp.next
        return
q = Queue()

for i in range(0,10):
    q.enqueue(i)

```

```python
q.traverse()

print(f"Front (peek): {q.peek()}")       # 10
print(f"Dequeued: {q.dequeue()}")        # 10
print(f"Dequeued: {q.dequeue()}")        # 20
print(f"New Front: {q.peek()}")          # 30
print(f"Height: {q.height}")             # 3

print("\nQueue now:")
q.traverse()
# Output: 30 -> 40 -> 50 -> None
```
