```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedList:
    def __init__(self,value):
        # we created 2 pointer
        new_node = Node(value)
        self.head = new_node
        self.tail= new_node
        self.n = 1 

    def print_list(self):
        temp = self.head
        while temp is not None:
            print(temp.value)
            temp = temp.next

    def append(self,value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else: 
            self.tail.next = new_node
            self.tail = new_node    
        self.n +=1
                   
    def pop(self):
        if self.n == 0:
            return None
        temp = self.head
        pre= self.head

        # 11->5->4->None
        while (temp.next):
            pre = temp
            temp = temp.next
            # when temp = None break
        # then, self.tail value will be pre means -1 so pre =5 
        self.tail = pre 
        self.tail.next = None
        self.n -=1
        if self.n == 0:
            self.head = self.tail = None

    def prepend(self,value):
        new_node = Node(value)
        if self.n == 0:
            self.head = self.tail= new_node
        else:
            new_node.next = self.head
            self.head = new_node
        self.n +=1
        return True

    def pop_first(self):
        if self.n == 0:
            return None
        temp = self.head
        temp.next = None
        self.n -=1
        if self.n ==0:
            self.tail=None
        return True

    # index value
    def get(self,index):
        if index <0 or index >= self.n:
            return None
        temp = self.head
        for _ in range(index):
            temp = temp.next
        return temp

    # replace value at index in linkedlist
    def set_value(self,index,value):
        temp = self.get(index)
        if temp:
            temp.value = value
            return True
        return False

    # insert_at() in linkedlist
    def insert(self,index,value):
        if index < 0 or index > self.n:
            return None
        if index == 0:
            return self.prepend(value)
        if index == self.n:
            return self.append(value)

        new_node = Node(value)
        temp = self.get(index -1 )
        new_node.next = temp.next
        temp.next = new_node
        self.n +=1
        return True

    def remove(self,index):
        if index < 0 or index > self.n:
            return None
        if index == 0:
            return self.pop_first()
        if index ==self.n:
            return self.pop()

        # lets say index  5 -1= 4 
        pre = self.get(index-1)
        # now we get pre value is 4 and pre.next = 5 
        temp = pre.next
        #  pre.next 5 == temp.next 6
        pre.next = temp.next 
        self.n -=1
        return temp

    def reverse(self):
        if self.n == 0:
            return None

        temp = self.head
        self.head = self.tail 
        self.tail = temp 
        after = temp.next 
        before = None

        for _ in range(self.n):
            after =temp.next
            before = temp
            temp = after 

        return True 
    

        
    

        
       
```
